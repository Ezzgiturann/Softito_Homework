---## Bölüm 6 — Filtreleme ve Yeni Sütun**Soru 6.1:** Toplam tutarı 300 TL'nin üzerinde olan siparişleri filtreleyin.**Soru 6.2:** Ürün adedi 2'den fazla VE toplam tutar 200 TL'den fazla olan siparişleri bulun (`&` kullanın).**Soru 6.3:** `birim_fiyat` adında yeni bir sütun ekleyin; değeri `toplam_tutar / urun_adedi` olsun.**Soru 6.4:** `apply` ve `lambda` kullanarak kargo süresi 3 günden fazlaysa "Yavaş", değilse "Hızlı" yazan bir `kargo_durumu` sütunu oluşturun.

```python
# 6.1 Toplam tutarı 300 TL üzerinde olanlar
print("300 TL Üstü Siparişler:\n", df[df["toplam_tutar"] > 300])
```

```python
# 6.2 Ürün adedi > 2 VE toplam tutar > 200
print("\nÜrün Adedi > 2 ve Tutar > 200:\n", df[(df["urun_adedi"] > 2) & (df["toplam_tutar"] > 200)])
```

```python
# 6.3 'birim_fiyat' sütunu ekleme
df["birim_fiyat"] = df["toplam_tutar"] / df["urun_adedi"]
```

```python
# 6.4 apply ve lambda ile 'kargo_durumu' sütunu
df["kargo_durumu"] = df["kargo_suresi"].apply(lambda x: "Yavaş" if x > 3 else "Hızlı")
```

---## Bölüm 9 — Merge (Tabloları Birleştirme)Aşağıdaki iki tabloyu kullanın:

```python
musteriler = pd.DataFrame({
    "musteri_id": [1, 2, 3, 4],
    "ad": ["Baran", "Ceylin", "Emre", "Fulya"]
})

siparisler = pd.DataFrame({
    "musteri_id": [1, 2, 5],
    "urun": ["Kulaklık", "Şarj Kablosu", "Powerbank"],
    "tutar": [450, 90, 320]
})
```

**Soru 9.1:** `musteriler` ve `siparisler` tablolarını **inner join** ile birleştirin — kaç müşterinin siparişi eşleşiyor?**Soru 9.2:** Aynı tabloları **left join** ile birleştirin — hiç siparişi olmayan müşteri(ler) sonuçta ne görünüyor?**Soru 9.3:** Aynı tabloları **outer join** ile birleştirip, hem eşleşmeyen müşterileri hem eşleşmeyen siparişleri gözlemleyin.

```python
# 9.1 Inner Join (Ortak müşteri_id'ler: 1 ve 2 eşleşir -> 2 sipariş)
inner_df = pd.merge(musteriler, siparisler, on="musteri_id", how="inner")
print("--- Inner Join ---\n", inner_df)
```

```python
# 9.2 Left Join (Siparişi olmayan 3 ve 4 nolu müşterilerin ürün/tutar kısımları NaN gelir)
left_df = pd.merge(musteriler, siparisler, on="musteri_id", how="left")
print("\n--- Left Join ---\n", left_df)
```

```python
# 9.3 Outer Join (Her iki tablodaki tüm kayıtlar birleşir, eşleşmeyenler NaN olur)
outer_df = pd.merge(musteriler, siparisler, on="musteri_id", how="outer")
print("\n--- Outer Join ---\n", outer_df)
```

---## Bölüm 10 — Concat**Soru 10.1:** `hafta1 = pd.DataFrame({"gun":["Pzt","Sal"], "adet":[20,35]})` ve `hafta2 = pd.DataFrame({"gun":["Çar","Per"], "adet":[28,40]})` tablolarını `ignore_index=True` ile alt alta birleştirin.**Soru 10.2:** `gelir = pd.DataFrame({"toplam_gelir":[1000,1750]})` tablosunu, `hafta1` tablosuyla `axis=1` kullanarak yan yana birleştirin.

```python
hafta1 = pd.DataFrame({"gun": ["Pzt", "Sal"], "adet": [20, 35]})
hafta2 = pd.DataFrame({"gun": ["Çar", "Per"], "adet": [28, 40]})
gelir = pd.DataFrame({"toplam_gelir": [1000, 1750]})
```

```python
# 10.1 Alt alta birleştirme (ignore_index=True)
df_alt_alta = pd.concat([hafta1, hafta2], ignore_index=True)
print("--- Alt Alta Concat ---\n", df_alt_alta)
```

```python
# 10.2 Yan yana birleştirme (axis=1)
df_yan_yana = pd.concat([hafta1, gelir], axis=1)
print("\n--- Yan Yana Concat (axis=1) ---\n", df_yan_yana)
```

---## Bölüm 11 — GroupByAşağıdaki veriyi kullanın:

```python
df_calisan = pd.DataFrame({    "sube": ["Kadıköy", "Beşiktaş", "Kadıköy", "Şişli", "Beşiktaş"],    "ad": ["Baran", "Ceylin", "Emre", "Fulya", "Gökhan"],    "satis": [12000, 18500, 9500, 21000, 16000]})
```

**Soru 11.1:** Şubeye göre **toplam** satışı bulun.**Soru 11.2:** Şubeye göre **ortalama** satışı bulun.**Soru 11.3:** `.agg(["sum","mean","count"])` kullanarak şube başına üç istatistiği aynı anda hesaplayın.

```python
# 11.2 Şube bazında ortalama satış
print("\nŞube Ortalama Satış:\n", df_calisan.groupby("sube")["satis"].mean())
```

```python
# 11.3 .agg(["sum", "mean", "count"]) ile çoklu istatistik
print("\nŞube Özet Tablosu (.agg):\n", df_calisan.groupby("sube")["satis"].agg(["sum", "mean", "count"]))
```

---## Bölüm 12 — Pivot TableAşağıdaki veriyi kullanın:

```python
df_satis = pd.DataFrame({    "sehir": ["Bursa", "Adana", "Antalya", "Bursa", "Adana"],    "kategori": ["Elektronik", "Giyim", "Elektronik", "Giyim", "Elektronik"],    "satis": [5000, 3200, 4100, 2800, 3900]})
```

**Soru 12.1:** `index="sehir"`, `columns="kategori"`, `values="satis"` ile bir pivot tablo oluşturun.**Soru 12.2:** Aynı pivot tabloyu `aggfunc="sum"` ve `fill_value=0` parametreleriyle oluşturup eksik hücrelerin nasıl doldurulduğunu gözlemleyin.

```python
# 12.1 Temel Pivot Table
pivot1 = pd.pivot_table(df_satis, index="sehir", columns="kategori", values="satis")
print("--- Pivot Tablo 1 ---\n", pivot1)

```

```python
# 12.2 aggfunc='sum' ve fill_value=0 parametreleriyle
pivot2 = pd.pivot_table(df_satis, index="sehir", columns="kategori", values="satis", aggfunc="sum", fill_value=0)
print("\n--- Boş Değerleri 0 Olan Pivot Tablo ---\n", pivot2)
```

---## Bölüm 15 — Matplotlib ile Grafik ÇizmeAşağıdaki hücreyi çalıştırarak bir yıllık çeyreklik gelir verisi oluşturun:

```python
df_gelir = pd.DataFrame({    "ceyrek": ["Ç1", "Ç2", "Ç3", "Ç4"],    "gelir": [42000, 55000, 48000, 61000],    "gider": [30000, 38000, 33000, 40000]})
```

**Soru 15.1:** `gelir` değerlerini `kind="line"` ile, üzerinde nokta işaretleri (`marker="o"`) olacak şekilde çizgi grafikle çizin; başlık ve eksen etiketleri ekleyin.**Soru 15.2:** `gelir` değerlerini çeyreklere göre `kind="bar"` ile bar grafikle çizin (renk seçiminiz size ait).**Soru 15.3:** `gelir` sütununu çeyreklerin payını gösterecek şekilde `kind="pie"` ile pasta grafikle çizin (`autopct="%1.1f%%"` kullanın).**Soru 15.4:** `gelir` ve `gider` sütunlarını `kind="scatter"` ile scatter grafikte karşılaştırın (x ekseni gelir, y ekseni gider olsun) ve grafiğin ne anlattığını bir cümleyle yorumlayın.

```python
# 15.1 Çizgi Grafiği (Line Plot)
plt.figure(figsize=(6, 3))
plt.plot(df_gelir["ceyrek"], df_gelir["gelir"], marker="o", color="blue", linewidth=2)
plt.title("Çeyreklik Gelir Grafiği")
plt.xlabel("Çeyrek")
plt.ylabel("Gelir (TL)")
plt.grid(True)
plt.show()
```

```python
# 15.2 Bar Grafiği (Bar Plot)
plt.figure(figsize=(6, 3))
plt.bar(df_gelir["ceyrek"], df_gelir["gelir"], color="teal")
plt.title("Çeyreklere Göre Gelir Dağılımı")
plt.xlabel("Çeyrek")
plt.ylabel("Gelir (TL)")
plt.show()
```

```python
# 15.3 Pasta Grafiği (Pie Chart)
plt.figure(figsize=(5, 5))
plt.pie(df_gelir["gelir"], labels=df_gelir["ceyrek"], autopct="%1.1f%%", startangle=90)
plt.title("Gelirin Çeyreklere Göre Payı")
plt.show()
```

```python
# 15.4 Dağılım Grafiği (Scatter Plot) ve Yorum
plt.figure(figsize=(6, 3))
plt.scatter(df_gelir["gelir"], df_gelir["gider"], color="crimson", s=80)
plt.title("Gelir - Gider Karşılaştırması")
plt.xlabel("Gelir (TL)")
plt.ylabel("Gider (TL)")
plt.grid(True)
plt.show()
```

```python
print("Yorum: Gelir arttıkça giderin de düzenli olarak arttığı doğrusal (pozitif yönlü) bir ilişki görülmektedir.")
```

En çok satan 3 kursu `.sort_values()` ile bulun.4. Kategori bazında toplam satışı gösteren bir bar grafik çizin.5. `apply` ile satış adedi 50 üzeri olanlara "Popüler", altındakilere "Standart" yazan yeni bir sütun ekleyin.

```python
# 1. En az 8 satırlık online kurs veri seti
df_kurs = pd.DataFrame({
    "kurs_adi": [
        "Python Sıfırdan İleriye", "UI/UX Tasarım Temelleri", "Dijital Pazarlama Uzmanlığı",
        "React ile Web Geliştirme", "Figma ile Arayüz Tasarımı", "SEO ve İçerik Pazarlaması",
        "Makine Öğrenmesi 101", "Veri Analizi Kampı"
    ],
    "kategori": ["Yazılım", "Tasarım", "Pazarlama", "Yazılım", "Tasarım", "Pazarlama", "Yazılım", "Yazılım"],
    "satis_adedi": [85, 42, 38, 64, 55, 30, 92, 78]
})
```

```python
print("--- Kurs Veri Seti ---")
display(df_kurs)
```

```python
# 2. Kategori bazında toplam satış adedi (groupby)
kategori_satis = df_kurs.groupby("kategori")["satis_adedi"].sum()
print("\n--- Kategori Bazında Toplam Satış ---\n", kategori_satis)
```

```python
# 3. En çok satan 3 kurs (.sort_values)
en_cok_satanlar = df_kurs.sort_values("satis_adedi", ascending=False).head(3)
print("\n--- En Çok Satan 3 Kurs ---")
display(en_cok_satanlar)
```

```python
# 4. Kategori bazında toplam satışı gösteren bar grafik
plt.figure(figsize=(6, 3))
kategori_satis.plot(kind="bar", color="darkorange")
plt.title("Kategori Bazında Toplam Kurs Satışları")
plt.xlabel("Kategori")
plt.ylabel("Toplam Satış Adedi")
plt.xticks(rotation=0)
plt.show()
```

```python
# 5. apply ve lambda ile 'durum' sütunu ekleme (>50 Popüler, aksi Standart)
df_kurs["durum"] = df_kurs["satis_adedi"].apply(lambda x: "Popüler" if x > 50 else "Standart")
print("\n--- Güncel Kurs Tablosu ---")
display(df_kurs)
```
