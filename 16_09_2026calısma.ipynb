```python
#calısma1
#marka, model ve hiz (km/s) bilgilerini tutan bir Araba sınıfı oluşturun.
#hızlandır(artis) metodu ile hızı artırın.
#bilgi_goster() metodu ile araç bilgilerini yazdırın.
# ==========================================
# 1. ARABA SINIFI
# ==========================================
class Araba:
    def __init__(self, marka, model, hiz=0):
        self.marka = marka
        self.model = model
        self.hiz = hiz

    def hizlandir(self, artis):
        self.hiz += artis
        print(f"{self.marka} {self.model} {artis} km/s hızlandı. Güncel hız: {self.hiz} km/s")

    def bilgi_goster(self):
        print(f"Marka: {self.marka}, Model: {self.model}, Hız: {self.hiz} km/s")



#calısma2
#Dikdörtgen Sınıfı
#uzunluk ve genislik bilgilerini tutan bir Dikdortgen sınıfı oluşturun.
#alan_hesapla() metodu ile alanı hesaplayın (uzunluk * genislik).
#cevre_hesapla() metodu ile çevreyi hesaplayın (2 * (uzunluk + genislik)).
# ==========================================
# 2. DİKDÖRTGEN SINIFI
# ==========================================
class Dikdortgen:
    def __init__(self, uzunluk, genislik):
        self.uzunluk = uzunluk
        self.genislik = genislik

    def alan_hesapla(self):
        return self.uzunluk * self.genislik

    def cevre_hesapla(self):
        return 2 * (self.uzunluk + self.genislik)




#calısma3
#Kitap Sınıfı
#ad, yazar ve sayfa_sayisi bilgilerini tutan bir Kitap sınıfı oluşturun.
#ozet() metodu ile "Suç ve Ceza - Dostoyevski (671 sayfa)"
#formatında bilgi döndürün.
# ==========================================
# 3. KİTAP SINIFI
# ==========================================
class Kitap:
    def __init__(self, ad, yazar, sayfa_sayisi):
        self.ad = ad
        self.yazar = yazar
        self.sayfa_sayisi = sayfa_sayisi

    def ozet(self):
        return f"{self.ad} - {self.yazar} ({self.sayfa_sayisi} sayfa)"


# ==========================================
# TEST KODLARI (Sınıfların Kullanımı)
# ==========================================
if __name__ == "__main__":
    print("--- Araba Testi ---")
    araba1 = Araba("Toyota", "Corolla", 50)
    araba1.bilgi_goster()
    araba1.hizlandir(30)
    araba1.bilgi_goster()

    print("\n--- Dikdörtgen Testi ---")
    dikdortgen1 = Dikdortgen(10, 5)
    print(f"Alan: {dikdortgen1.alan_hesapla()}")
    print(f"Çevre: {dikdortgen1.cevre_hesapla()}")

    print("\n--- Kitap Testi ---")
    kitap1 = Kitap("Suç ve Ceza", "Dostoyevski", 671)
    print(kitap1.ozet())
```
