```python
calisan_verisi = {
    "isim": ["Ahmet", "Buse", "Cem", "Derya", "Emre", "Fatma"],
    "yas": [28, 24, 35, 19, 42, 23],
    "sehir": ["İstanbul", "Ankara", "İzmir", "İstanbul", "Bursa", "Ankara"],
    "maas": [28000, 22000, 45000, 18000, 52000, 24000]
}
```

```python

df_calisan = pd.DataFrame(calisan_verisi)
```

```python
print("\nİlk 3 Satır (.head):\n", df_calisan.head(3))
```

```python
print("\nSon 2 Satır (.tail):\n", df_calisan.tail(2))
```

```python
print("\nBoyut (.shape):", df_calisan.shape)
```

```python
print("\nVeri Tipi Bilgileri (.info):")
```

```python
df_calisan.info()
```

```python
print("\nSayısal Özet (.describe):\n", df_calisan.describe())
```

```python
# 6. SÜTUN VE SATIR SEÇME

print("\nSadece 'isim' sütunu:\n", df_calisan["isim"])
print("\n'isim' ve 'yas' sütunları:\n", df_calisan[["isim", "yas"]])
```

```python
# .loc[] ile ilk 3 satır (0, 1, 2 indexleri dahil)
print("\n.loc ile ilk 3 satır:\n", df_calisan.loc[0:2])
```

```python
# .iloc[] ile 2. satır ve 2. sütunun kesiştiği hücre (0-indeksli: satir 2, sutun 2 -> 'İzmir')
print("\n.iloc[2, 2] değeri:", df_calisan.iloc[2, 2])
```

```python
#7) FİLTRELEME (KOŞULLU SEÇİM)
```

```python
print("\nYaşı 25'ten büyük olanlar:\n", df_calisan[df_calisan["yas"] > 25])





```

```python
print("\nMaaşı 20000 - 30000 arasında olanlar (&):\n",
      df_calisan[(df_calisan["maas"] >= 20000) & (df_calisan["maas"] <= 30000)])
```

```python
print("\nİstanbul'da yaşayanlar:\n", df_calisan[df_calisan["sehir"] == "İstanbul"])
```

```python
print("\nYaşı 20'den küçük VEYA 40'tan büyük olanlar (|):\n",
      df_calisan[(df_calisan["yas"] < 20) | (df_calisan["yas"] > 40)])
```

```python
#8)YENİ SÜTUN EKLEME VE GÜNCELLEME
```

```python
# 2026 yılına göre doğum yılı sütunu
df_calisan["dogum_yili"] = 2026 - df_calisan["yas"]
```

```python
# Maaş sütununu %10 artırma
df_calisan["maas"] = df_calisan["maas"] * 1.10
```

```python
# apply() ve lambda ile yaş durumu sütunu
df_calisan["yas_grubu"] = df_calisan["yas"].apply(lambda x: "Yetişkin" if x >= 18 else "Çocuk")
```

```python
print("\nGüncellenmiş Tablo:\n", df_calisan)
```

```python
# 9. SIRALAMA

print("\nYaşa Göre Küçükten Büyüğe Sıralama:\n",
      df_calisan.sort_values("yas"))
```

```python
print("\nMaaşa Göre Büyükten Küçüğe Sıralama:\n",
      df_calisan.sort_values("maas", ascending=False))
```

```python
print("\nÖnce Şehir, Sonra Yaşa Göre Sıralama:\n",
      df_calisan.sort_values(["sehir", "yas"]))
```

```python
# BONUS MİNİ ÖDEV
bonus_verisi = {
    "isim": ["Ece", "Barış", "Selin", "Kaan", "Ezgi", "Onur"],
    "yas": [22, 24, 21, 25, 20, 23],
    "sehir": ["İzmir", "Ankara", "İstanbul", "Bursa", "İzmir", "Ankara"],
    "not1": [85, 60, 90, 65, 75, 50],
    "not2": [90, 70, 95, 60, 80, 55],
    "not3": [80, 65, 85, 70, 70, 60]
}
```

```python
df_bonus = pd.DataFrame(bonus_verisi)
```

```python
# 3 dersin ortalamasını yeni sütun olarak hesaplama
df_bonus["ortalama"] = (df_bonus["not1"] + df_bonus["not2"] + df_bonus["not3"]) / 3
```

```python
# Ortalaması 70'in üzerinde olanları filtreleme ve yaşa göre sıralama
sonuc = df_bonus[df_bonus["ortalama"] > 70].sort_values("yas")
```

```python
print("\n--- Bonus Görev Sonucu (Ortalaması > 70, Yaşa Göre Sıralı) ---")
```

```python
print(sonuc)
```
